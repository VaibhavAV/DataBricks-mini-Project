
READ And Process Data

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("CustomerDataProcessing").getOrCreate()

spark 
```

```python
df = spark.read.format("csv").option("header", "true").load("/Volumes/workspace/default/databricks/customers.csv")
```

```python
df.count()
```

```python
#Select specific columns

df.select('customer_id','city','country').show(5)
```

```python
df.show(5)
```

```python
df.printSchema()
```

```python
# Filter  = show only active customers
df.filter(df.is_active == True).show(5)
```

```python
# Sorting data using orderby
df.orderBy('customer_id').show(5)
```

```python
# Combine sorting and limiting = Show me the 10 customers with the highest customer IDs.

df.orderBy(
    df.customer_id.desc()
).limit(10).show(5)
```

```python
# distinct() – Get unique records
df.distinct().count()

#unique cities 
df.select("customer_id").distinct().count()
```

```python
from pyspark.sql.functions import   *

df= df.withColumn('registration_date', to_date(col('registration_date'),'yyyy-MM-dd'))  \
    .withColumn('is_active', col("is_active").cast("boolean"))
```

```python
df.show(5)
df.printSchema()
```

```python
# filtering recent customers

df_recent_customers = df.filter(col('registration_date') >= lit('2023-07-01'))
df_recent_customers.show(5) 
```

Joining And Analyzing Customers and orders


```python
orders_df = spark.read.format("csv").option("header", 'true').option("inferSchema", True).load("/Volumes/workspace/default/databricks/orders.csv")
#orders_df.show(5)
```

```python
orders_df.printSchema()
```

```python
#calculate avg amount
from pyspark.sql.functions import avg
orders_df.select(avg('total_amount')).show() 

#Min max values
from pyspark.sql.functions import min, max
orders_df.select(min('total_amount'),max('total_amount')).show()
```

```python
#Finding deliverd orders


deliverd_orders = orders_df.filter(orders_df.status == "Delivered")
deliverd_orders.count()

pending_orders = orders_df.filter( orders_df.status == "Pending")
pending_orders.count()
```

```python
#Find all orders with total_amount greater than 500.
orders_df.filter( orders_df.total_amount > 500).count()
```
