```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Orders_df").getOrCreate()
spark
```

```python
order_df = spark.read.format("csv").option('header','true').option('inferSchema',True).load('/Volumes/workspace/default/databricks/orders.csv')
order_df.show(5)
```

```python
#Learn the aggregation functions
from pyspark.sql.functions import count, sum, avg, min, max

#Count total orders
order_df.select(count('*').alias('total_orders')).show()

```

```python
#Calculate total sales
order_df.select( sum('total_amount').alias("total_scale")).show()

#Calculate average, minimum and maximum order value

order_df.select( avg('total_amount').alias("total_avg"), min('total_amount').alias("min"), max('total_amount').alias("max")).show()
```

```python
# Group by
order_df.groupBy("status").count().show()

#GroupBy with multiple aggregations
order_df.groupBy('status').agg(
    count("*").alias('total_orders'),
    sum('total_amount').alias("total_sales"),
    avg('total_amount').alias("avg_sales"),   
).show()
```

```python
#Sort the aggregated results

order_df.groupBy('status').agg(
    sum('total_amount').alias('total_sales')
).orderBy('total_sales', ascending=False).show()
```
